```python
import pandas as pd
import plotly.express as px

# -------------------------------------------------------------
file_path = r'C:\Users\Acer\Downloads\3_Disease_Analytics.xlsx'
# -------------------------------------------------------------

# Data Load Karein
df_dis = pd.read_excel(file_path, sheet_name='Raw Data')

# 1. Total Patients per Disease (Bar Chart)
dis_counts = df_dis['Disease'].value_counts().reset_index()
dis_counts.columns = ['Disease', 'Patient Count']

fig_dis = px.bar(
    dis_counts, 
    x='Disease', 
    y='Patient Count',
    color='Disease',
    title='<b>1. Top Diseases by Patient Inflow</b>',
    text_auto=True,
    color_discrete_sequence=px.colors.qualitative.Dark2
)
fig_dis.update_layout(showlegend=False, template='plotly_white')
fig_dis.show()

# 2. Disease vs Admission Status / Severity (Stacked Bar Chart)
dis_status = df_dis.groupby(['Disease', 'Admission Status']).size().reset_index(name='Count')

fig_severity = px.bar(
    dis_status, 
    x='Disease', 
    y='Count', 
    color='Admission Status',
    title='<b>2. Patient Severity (Critical vs Normal) per Disease</b>',
    text_auto=True,
    color_discrete_sequence=px.colors.qualitative.Set1
)
fig_severity.update_layout(template='plotly_white')
fig_severity.show()

# 3. Disease Spread by Age Group (Heatmap)
dis_age = pd.crosstab(df_dis['Disease'], df_dis['Age Group'])

fig_heat = px.imshow(
    dis_age,
    labels=dict(x="Age Group", y="Disease", color="Patients"),
    x=dis_age.columns,
    y=dis_age.index,
    title='<b>3. Disease Distribution Across Age Groups</b>',
    text_auto=True,
    color_continuous_scale='Reds'
)
fig_heat.show()
```
