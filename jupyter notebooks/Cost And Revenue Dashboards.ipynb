```python
import pandas as pd
import plotly.express as px

# -------------------------------------------------------------
# AAPKO BUS YAHAN APNA EXACT FILE PATH COPY-PASTE KARNA HAI
# -------------------------------------------------------------
file_path = r'C:\Users\Acer\Downloads\5_Cost_and_Revenue.xlsx'
# -------------------------------------------------------------

# Data Load Karein
df_cost = pd.read_excel(file_path, sheet_name='Raw Data')

# 1. Revenue Share by Payment Mode (Donut Chart)
pay_rev = df_cost.groupby('Payment Mode')['Treatment Cost (INR)'].sum().reset_index()
pay_rev.columns = ['Payment Mode', 'Total Revenue']

fig_pay = px.pie(
    pay_rev, 
    names='Payment Mode', 
    values='Total Revenue',
    title='<b>1. Revenue Share by Payment Mode (Insurance vs Cash vs UPI)</b>',
    hole=0.45,
    color_discrete_sequence=px.colors.qualitative.Pastel1
)
fig_pay.update_traces(textinfo='percent+label')
fig_pay.show()

# 2. Total Revenue Generated by Disease (Bar Chart)
dis_rev = df_cost.groupby('Disease')['Treatment Cost (INR)'].sum().reset_index()
dis_rev = dis_rev.sort_values(by='Treatment Cost (INR)', ascending=False)

fig_dis_r = px.bar(
    dis_rev, 
    x='Disease', 
    y='Treatment Cost (INR)',
    color='Disease',
    title='<b>2. Total Hospital Revenue Generated per Disease</b>',
    text_auto='.2s',
    color_discrete_sequence=px.colors.qualitative.Set2
)
fig_dis_r.update_layout(showlegend=False, template='plotly_white')
fig_dis_r.show()

# 3. Treatment Cost Distribution across Wards (Box Plot)
fig_box = px.box(
    df_cost, 
    x='Ward Type', 
    y='Treatment Cost (INR)', 
    color='Ward Type',
    title='<b>3. Treatment Cost Range & Outliers by Ward Type</b>',
    color_discrete_sequence=px.colors.qualitative.Bold
)
fig_box.update_layout(template='plotly_white')
fig_box.show()
```
