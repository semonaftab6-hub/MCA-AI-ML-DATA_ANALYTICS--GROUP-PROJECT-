```python
import pandas as pd
```

```python
# Create all 5 MySQL tables for Healthcare Analytics
tables_creation_queries = [
    # Table 1: Hospital Operations
    """
    CREATE TABLE IF NOT EXISTS hospital_operations (
        patient_id VARCHAR(20) PRIMARY KEY,
        admission_date DATE,
        discharge_date DATE,
        length_of_stay INT,
        ward_type VARCHAR(50),
        admission_status VARCHAR(50)
    )
    """,
    # Table 2: Patient Analytics
    """
    CREATE TABLE IF NOT EXISTS patient_analytics (
        patient_id VARCHAR(20) PRIMARY KEY,
        age INT,
        age_group VARCHAR(30),
        gender VARCHAR(20),
        city VARCHAR(50)
    )
    """,
    # Table 3: Disease Analytics
    """
    CREATE TABLE IF NOT EXISTS disease_analytics (
        patient_id VARCHAR(20) PRIMARY KEY,
        disease VARCHAR(100),
        admission_status VARCHAR(50),
        age_group VARCHAR(30),
        gender VARCHAR(20)
    )
    """,
    # Table 4: Doctor Performance
    """
    CREATE TABLE IF NOT EXISTS doctor_performance (
        patient_id VARCHAR(20),
        doctor_name VARCHAR(100),
        disease VARCHAR(100),
        ward_type VARCHAR(50),
        treatment_cost DECIMAL(12,2)
    )
    """,
    # Table 5: Cost and Revenue (Your assigned module)
    """
    CREATE TABLE IF NOT EXISTS cost_and_revenue (
        patient_id VARCHAR(20),
        disease VARCHAR(100),
        ward_type VARCHAR(50),
        treatment_cost DECIMAL(12,2),
        payment_mode VARCHAR(50)
    )
    """,
]

# Execute table creation queries
for query in tables_creation_queries:
  cursor.execute(query)

print("All 5 Tables Created Successfully!")
```

```python
import pandas as pd

# Load and insert data for Hospital Operations
df1 = pd.read_excel("1_Hospital_Operation (1).xlsx")

for _, row in df1.iterrows():
  cursor.execute(
      """
        INSERT IGNORE INTO hospital_operations 
        (patient_id, admission_date, discharge_date, length_of_stay, ward_type, admission_status)
        VALUES (%s, %s, %s, %s, %s, %s)
    """,
      (
          row["Patient ID"],
          row["Admission Date"],
          row["Discharge Date"],
          row["Length of Stay (Days)"],
          row["Ward Type"],
          row["Admission Status"],
      ),
  )

conn.commit()
print("Hospital Operations data inserted successfully!")
```

```python
# Load and insert data for Patient Analytics
df2 = pd.read_excel("2_Patient_Analytics.xlsx")

for _, row in df2.iterrows():
  cursor.execute(
      """
        INSERT IGNORE INTO patient_analytics 
        (patient_id, age, age_group, gender, city)
        VALUES (%s, %s, %s, %s, %s)
    """,
      (
          row["Patient ID"],
          row["Age"],
          row["Age Group"],
          row["Gender"],
          row["City"],
      ),
  )

conn.commit()
print("Patient Analytics data inserted successfully!")
```

```python
# Load and insert data for Disease Analytics
df3 = pd.read_excel("3_Disease_Analytics.xlsx")

for _, row in df3.iterrows():
  cursor.execute(
      """
        INSERT IGNORE INTO disease_analytics 
        (patient_id, disease, admission_status, age_group, gender)
        VALUES (%s, %s, %s, %s, %s)
    """,
      (
          row["Patient ID"],
          row["Disease"],
          row["Admission Status"],
          row["Age Group"],
          row["Gender"],
      ),
  )

conn.commit()
print("Disease Analytics data inserted successfully!")
```

```python
# Load and insert data for Doctor Performance
df4 = pd.read_excel("4_Doctor_Performance.xlsx")

for _, row in df4.iterrows():
  cursor.execute(
      """
        INSERT INTO doctor_performance 
        (patient_id, doctor_name, disease, ward_type, treatment_cost)
        VALUES (%s, %s, %s, %s, %s)
    """,
      (
          row["Patient ID"],
          row["Doctor Name"],
          row["Disease"],
          row["Ward Type"],
          row["Treatment Cost (INR)"],
      ),
  )

conn.commit()
print("Doctor Performance data inserted successfully!")
```

```python
# Load and insert data for Cost and Revenue (Main Module)
df5 = pd.read_excel("5_Cost_and_Revenue.xlsx")

for _, row in df5.iterrows():
  cursor.execute(
      """
        INSERT INTO cost_and_revenue 
        (patient_id, disease, ward_type, treatment_cost, payment_mode)
        VALUES (%s, %s, %s, %s, %s)
    """,
      (
          row["Patient ID"],
          row["Disease"],
          row["Ward Type"],
          row["Treatment Cost (INR)"],
          row["Payment Mode"],
      ),
  )

conn.commit()
print("Cost and Revenue data inserted successfully!")
```
