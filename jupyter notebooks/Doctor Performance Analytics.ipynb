```python
import pandas as pd
import plotly.express as px

# -------------------------------------------------------------
# AAPKO BUS YAHAN APNA EXACT FILE PATH COPY-PASTE KARNA HAI
# -------------------------------------------------------------
file_path = r'C:\Users\Acer\Downloads\4_Doctor_Performance.xlsx'
# -------------------------------------------------------------

# Data Load Karein
df_doc = pd.read_excel(file_path, sheet_name='Raw Data')

# 1. Total Patients Treated per Doctor (Bar Chart)
doc_patients = df_doc['Doctor Name'].value_counts().reset_index()
doc_patients.columns = ['Doctor Name', 'Patient Count']

fig_doc_p = px.bar(
    doc_patients, 
    x='Doctor Name', 
    y='Patient Count',
    color='Doctor Name',
    title='<b>1. Total Patient Load per Doctor</b>',
    text_auto=True,
    color_discrete_sequence=px.colors.qualitative.Pastel
)
fig_doc_p.update_layout(showlegend=False, template='plotly_white')
fig_doc_p.show()

# 2. Total Revenue Generated per Doctor (Horizontal Bar Chart)
doc_rev = df_doc.groupby('Doctor Name')['Treatment Cost (INR)'].sum().reset_index()
doc_rev = doc_rev.sort_values(by='Treatment Cost (INR)', ascending=True)

fig_doc_r = px.bar(
    doc_rev, 
    x='Treatment Cost (INR)', 
    y='Doctor Name',
    orientation='h',
    color='Treatment Cost (INR)',
    title='<b>2. Total Revenue Generated (INR) per Doctor</b>',
    text_auto='.2s',
    color_continuous_scale='Viridis'
)
fig_doc_r.update_layout(template='plotly_white')
fig_doc_r.show()

# 3. Doctor-wise Patients Treated by Ward Type (Grouped Bar Chart)
doc_ward = df_doc.groupby(['Doctor Name', 'Ward Type']).size().reset_index(name='Patient Count')

fig_doc_w = px.bar(
    doc_ward, 
    x='Doctor Name', 
    y='Patient Count', 
    color='Ward Type',
    barmode='group',
    title='<b>3. Doctor Patient Distribution by Ward Type (ICU vs General vs Private)</b>',
    text_auto=True,
    color_discrete_sequence=px.colors.qualitative.Set2
)
fig_doc_w.update_layout(template='plotly_white')
fig_doc_w.show()
```
