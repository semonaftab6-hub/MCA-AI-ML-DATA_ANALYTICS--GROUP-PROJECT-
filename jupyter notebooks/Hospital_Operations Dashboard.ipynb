```python
import pandas as pd
import plotly.express as px

# -------------------------------------------------------------
file_path = r'C:\Users\Acer\Downloads\1_Hospital_Operation.xlsx'
# -------------------------------------------------------------

# File Read & Data Load
df_hosp = pd.read_excel(file_path, sheet_name='Raw Data')

# 1. Ward Utilization (Donut Chart)
ward_counts = df_hosp['Ward Type'].value_counts().reset_index()
ward_counts.columns = ['Ward Type', 'Patient Count']

fig_ward = px.pie(
    ward_counts, 
    names='Ward Type', 
    values='Patient Count',
    title='<b>1. Ward Utilization (ICU vs General vs Private)</b>',
    hole=0.45,
    color_discrete_sequence=px.colors.qualitative.Set2
)
fig_ward.update_traces(textinfo='percent+label')
fig_ward.show()

# 2. Admission Status (Bar Chart)
status_counts = df_hosp['Admission Status'].value_counts().reset_index()
status_counts.columns = ['Admission Status', 'Patient Count']

fig_status = px.bar(
    status_counts, 
    x='Admission Status', 
    y='Patient Count',
    color='Admission Status',
    title='<b>2. Patient Condition (Admission Status)</b>',
    text_auto=True,
    color_discrete_sequence=px.colors.qualitative.Pastel
)
fig_status.update_layout(showlegend=False, template='plotly_white')
fig_status.show()

# 3. Average Length of Stay by Ward (Bar Chart)
avg_stay_ward = df_hosp.groupby('Ward Type')['Length of Stay (Days)'].mean().reset_index()
avg_stay_ward['Length of Stay (Days)'] = avg_stay_ward['Length of Stay (Days)'].round(1)

fig_stay = px.bar(
    avg_stay_ward, 
    x='Ward Type', 
    y='Length of Stay (Days)',
    title='<b>3. Average Length of Stay (Days) by Ward Type</b>',
    text_auto=True,
    color='Ward Type',
    color_discrete_sequence=px.colors.qualitative.Dark2
)
fig_stay.update_layout(showlegend=False, template='plotly_white')
fig_stay.show()
```
