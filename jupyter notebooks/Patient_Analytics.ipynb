```python
import pandas as pd
import plotly.express as px

# -------------------------------------------------------------
file_path = r'C:\Users\Acer\Downloads\2_Patient_Analytics.xlsx'
# -------------------------------------------------------------

# File Read & Data Load
df_pat = pd.read_excel(file_path, sheet_name='Raw Data')

# 1. Gender Distribution (Donut Chart)
gender_counts = df_pat['Gender'].value_counts().reset_index()
gender_counts.columns = ['Gender', 'Patient Count']

fig_gender = px.pie(
    gender_counts, 
    names='Gender', 
    values='Patient Count',
    title='<b>1. Gender Distribution (Male vs Female)</b>',
    hole=0.45,
    color_discrete_sequence=['#1f77b4', '#e377c2']
)
fig_gender.update_traces(textinfo='percent+label')
fig_gender.show()

# 2. Age Group Distribution (Bar Chart)
age_counts = df_pat['Age Group'].value_counts().reset_index()
age_counts.columns = ['Age Group', 'Patient Count']

fig_age = px.bar(
    age_counts, 
    x='Age Group', 
    y='Patient Count',
    color='Age Group',
    title='<b>2. Patient Inflow by Age Group</b>',
    text_auto=True,
    color_discrete_sequence=px.colors.qualitative.Set2
)
fig_age.update_layout(showlegend=False, template='plotly_white')
fig_age.show()

# 3. City-wise Demographics (Horizontal Bar Chart)
city_counts = df_pat['City'].value_counts().reset_index()
city_counts.columns = ['City', 'Patient Count']

fig_city = px.bar(
    city_counts, 
    x='Patient Count', 
    y='City',
    orientation='h',
    color='Patient Count',
    title='<b>3. Top Patient Inflow by City</b>',
    color_continuous_scale='Blues'
)
fig_city.update_layout(template='plotly_white')
fig_city.show()
```
